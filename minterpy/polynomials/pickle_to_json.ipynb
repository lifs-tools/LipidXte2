```python
import glob
import pickle
import json_tricks as json
import os
```

```python
pickle_files = glob.glob(r'./*.pkl')
```

```python
for pickle_file in pickle_files:
    data = pickle.load(open(pickle_file,'rb'))

    json_file = os.path.join(os.path.basename(pickle_file).replace('.pkl', '.json'))
    
    with open(json_file, 'w') as f:
        json.dump(data, f)
```

```python
for pickle_file in pickle_files:
    filename = os.path.basename(pickle_file)
    filename = filename.replace('sn1FA', 'sn1_FA')
    filename = filename.replace('sn1CO2', 'sn1_CO2')
    filename = filename.replace('sn2FA', 'sn2_FA')
    filename = filename.replace('sn2CO2', 'sn2_CO2')
    filename = filename.replace('symFA', 'sym_FA')
    filename = filename.replace('symCO2', 'sym_CO2')

    tokens = filename.split('_')
    prefix = tokens[0] + '_' + tokens[1] + '_'
    sub = tokens[1]
    os.makedirs(sub, exist_ok=True)
    data = pickle.load(open(pickle_file,'rb'))

    json_file = os.path.join(sub, filename.replace(prefix, '').replace('.pkl', '.json'))
    
    with open(json_file, 'w') as f:
        json.dump(data, f)
```
